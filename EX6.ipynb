```python
demo1=np.array([1,2,3,4,5])
demo2=np.array([4,3,2,1,5])
res_map=np.array(list(map(lambda x,y:x+y,demo1,demo2)))
print("result of map using numpy",res_map)
```

```python
from functools import reduce
demo1=np.array([1,2,3,4,5])
demo2=np.array([4,3,2,1,5])
res_reduce=reduce(lambda x,y:x+y,demo2)
print("result of map using numpy",res_reduce)
```

```python
import pandas as pd
data={'A':[420,380,390],'B':[50,40,45]}
df=pd.DataFrame(data)
df['A_mapped']=df['A'].map(lambda x:x+2)
print("mapped values",df['A_mapped'])
```

```python
import pandas as pd
data={'A':[420,380,390],'B':[50,40,45]}
df=pd.DataFrame(data)
print(df)
df['sum']=df.apply(lambda row:row['A']+row['B'],axis=1)
print("sum of A & B using apply:",df['sum'])
```

```python
import pandas as pd
from functools import reduce
data={'A':[420,380,390],'B':[50,40,45]}
df=pd.DataFrame(data)
print(df)
res_reduce=reduce(lambda x,y:x+y,df['B'])
print(res_reduce)
```
