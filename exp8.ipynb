```python
df=pd.read_csv("CreditCard_DV.csv")
```

```python
import matplotlib.pyplot as plt
fig=plt.figure(figsize=[5,5])
ax=fig.add_axes([0,0,1,1])
ax.scatter(df['Credit_Limit'],df['avg_purchase'])
ax.set_tittle("credit limit vs avg purchase",fontsize=16)
ax.set_xlabel("credit limit",fontsize=12)
ax.set_ylabel("avg purchase",fontsize=12)
fig.show()
```
